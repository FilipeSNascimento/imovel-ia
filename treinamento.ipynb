```python
import matplotlib.pyplot as plt
import pandas as pd

serie = pd.Series(metricas['mae_validacao'])
ax = serie.plot.barh(color=['#8299B8', '#14B8A6'], title='Comparação de MAE na validação (menor é melhor)')
ax.set_xlabel('Erro absoluto médio (US$)')
plt.tight_layout()
plt.show()
```
